  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Cell ini memuat pustaka yang dibutuhkan. NumPy digunakan untuk perhitungan array, Pandas untuk membaca dan mengolah dataset, dan Matplotlib untuk membuat grafik. torch, torch.nn, dan torch.optim adalah komponen PyTorch untuk membangun model, fungsi loss, dan optimizer. TensorDataset dan DataLoader digunakan untuk membagi data menjadi mini-batch. Dua baris seed(42) mengunci angka acak agar hasil dapat diulang.</p>
</div>

```python
#202432080_AmeliaGrascella
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
np.random.seed(42)
torch.manual_seed(42)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Tahap pra-pemrosesan dilakukan dengan urutan berikut. (1) Konfigurasi: menentukan lokasi file, kolom tanggal (Date), kolom target (Close), panjang window 30, dan rasio data latih 80%. (2) Memuat data: file dibaca dengan read_csv, kolom tanggal diubah menjadi tipe datetime, lalu data diurutkan dari yang terlama karena urutan waktu wajib dijaga pada data time-series. (3) Eksplorasi: shape, head(), dan isnull().sum() menampilkan ukuran data, isi awal, dan jumlah nilai kosong; hasilnya 4.431 baris dan 7 kolom tanpa missing value. (4) Pembersihan: interpolate().bfill().ffill() disiapkan sebagai pengaman jika terdapat nilai kosong. (5) Pembagian data: 80% data awal digunakan untuk melatih dan 20% data akhir untuk menguji secara kronologis tanpa diacak, sehingga model tidak melihat data masa depan. (6) Normalisasi relatif per window: karena harga naik dari sekitar 50 hingga sekitar 2.800 sehingga data uji berada di luar rentang data latih, setiap window dibagi dengan harga pertamanya lalu dikurangi 1, sehingga model mempelajari perubahan relatif, bukan harga absolut. Harga acuan disimpan agar prediksi dapat dikembalikan ke harga asli. (7) Sliding window: fungsi create_sequences membentuk pasangan 30 hari sebelumnya sebagai input dan hari berikutnya sebagai target. Hasilnya 3.514 sampel latih dan 887 sampel uji, dan window data uji hanya berisi harga masa lalu sehingga tidak terjadi kebocoran data. (8) Data dikonversi menjadi tensor PyTorch dan data latih dibagi menjadi batch berisi 64 sampel.</p>
</div>

```python
#202432080_AmeliaGrascella
DATA_PATH = 'GOOGL.csv'
DATE_COL = 'Date'
TARGET_COL = 'Close'
WINDOW = 30
TRAIN_RATIO = 0.8
df = pd.read_csv(DATA_PATH)
df[DATE_COL] = pd.to_datetime(df[DATE_COL])
df = df.sort_values(DATE_COL).reset_index(drop=True)
print('Jumlah baris & kolom:', df.shape)
print(df.head())
print('Missing value:\n', df.isnull().sum())
df[TARGET_COL] = df[TARGET_COL].interpolate().bfill().ffill()
series = df[TARGET_COL].values.astype('float32')
split = int(len(series) * TRAIN_RATIO)
def create_sequences(data, window, start, end):
    X, y, base = [], [], []
    for i in range(start, end):
        w = data[i - window:i]
        b = w[0]
        X.append((w / b - 1).reshape(-1, 1))
        y.append(data[i] / b - 1)
        base.append(b)
    return np.array(X), np.array(y).reshape(-1, 1), np.array(base).reshape(-1, 1)
X_train_np, y_train_np, base_train = create_sequences(series, WINDOW, WINDOW, split)
X_test_np, y_test_np, base_test = create_sequences(series, WINDOW, split, len(series))
X_train = torch.tensor(X_train_np, dtype=torch.float32)
y_train = torch.tensor(y_train_np, dtype=torch.float32)
X_test = torch.tensor(X_test_np, dtype=torch.float32)
y_test = torch.tensor(y_test_np, dtype=torch.float32)
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)
print('Bentuk X_train:', X_train.shape, '| X_test:', X_test.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Kelas SequenceModel terdiri dari dua layer. nn.LSTM (1 input, 64 unit hidden, 1 layer, batch_first=True) memproses 30 hari berurutan dan menjalankan perhitungan forget gate, input gate, candidate cell state, dan output gate seperti pada Bab 2. nn.Linear mengubah hidden state 64 dimensi menjadi 1 nilai prediksi. Pada fungsi forward, hanya hidden state langkah terakhir (out[:, -1, :]) yang diteruskan ke layer Linear karena sudah merangkum informasi seluruh 30 hari.</p>
</div>

```python
#202432080_AmeliaGrascella
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=1, output_size=1):
        super(SequenceModel, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)
    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)
        out = self.fc(out[:, -1, :])
        return out
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Hyperparameter yang digunakan adalah 50 epoch, learning rate 0.01, dan hidden size 64. Fungsi loss yang dipakai adalah MSE yang sesuai untuk kasus regresi, dengan optimizer Adam. Pada setiap batch, zero_grad() mengosongkan gradien sebelumnya, model(xb) menghasilkan prediksi, criterion menghitung galat, loss.backward() menghitung gradien melalui backpropagation through time, dan optimizer.step() memperbarui bobot. Nilai loss tiap epoch disimpan pada train_losses untuk divisualisasikan dan dicetak setiap 10 epoch.</p>
</div>

```python
#202432080_AmeliaGrascella
epochs = 50
learning_rate = 0.01
hidden_size = 64
model = SequenceModel(input_size=1, hidden_size=hidden_size)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
train_losses = []
for epoch in range(epochs):
    model.train()
    epoch_loss = 0.0
    for xb, yb in train_loader:
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item() * xb.size(0)
    epoch_loss /= len(train_loader.dataset)
    train_losses.append(epoch_loss)
    if (epoch + 1) % 10 == 0:
        print(f'Epoch [{epoch + 1}/{epochs}] - Loss: {epoch_loss:.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Cell ini menggambar kurva loss terhadap epoch. Selanjutnya model diubah ke mode evaluasi (model.eval() dan torch.no_grad()) untuk memprediksi data uji. Hasil prediksi dan nilai aktual dikembalikan ke harga asli dengan rumus (nilai + 1) dikali harga acuan window, kemudian dihitung RMSE dan MAE. Grafik kedua membandingkan data aktual (garis tebal) dengan prediksi LSTM (garis putus-putus).</p>
</div>

```python
#202432080_AmeliaGrascella
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
model.eval()
with torch.no_grad():
    y_pred_scaled = model(X_test).numpy()
y_pred = (y_pred_scaled + 1) * base_test
y_true = (y_test_np + 1) * base_test
rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
mae = np.mean(np.abs(y_true - y_pred))
print(f'RMSE: {rmse:.4f} | MAE: {mae:.4f}')
plt.figure(figsize=(12, 5))
plt.plot(y_true, color='#114F6A', linewidth=1.8, label='Data Aktual')
plt.plot(y_pred, color='#1FA4B1', linewidth=1.8, linestyle='--', label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi LSTM pada Data Uji')
plt.xlabel('Indeks Waktu (Data Uji)')
plt.ylabel(TARGET_COL)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
